## 1. First command: `print()` and comments

`print()` outputs text or numbers to the screen just underneath the cell. Anything following a `#` on a line is a **comment** — Python ignores it, so we use comments to leave notes to ourselves inside the code.


```python
# This line is a comment. Python skips it.
print('hello world')

# print() can take multiple values, separated by commas:
print('2 + 2 =', 2 + 2)
```

## 2. Variables and types

A **variable** is a label that points to a value in memory. You assign one with `=` (a single equals sign). The three types we'll use constantly:

| Type | What it stores | Examples |
|---|---|---|
| `int` | whole numbers | `5`, `-3`, `100` |
| `float` | decimal numbers | `3.14`, `0.08`, `-1.5` |
| `str` | text in quotes | `'Toronto'`, `"St. Catharines"` |
| `bool` | True/False | `True`, `False` (capitalized) |

Use `type(x)` to check what Python thinks `x` is.


```python
a = 5                  # int
b = 10.0               # float (note the decimal)
name = 'Alex'          # str
is_student = True      # bool

print(a, b, name, is_student)
print(type(a), type(b), type(name), type(is_student))
```

## 3. Arithmetic

The seven arithmetic operators you'll actually use, with `a = 5, b = 10`:

| Operator | Example | Result | Meaning |
|---|---|---|---|
| `+` | `a + b` | `15` | addition |
| `-` | `a - b` | `-5` | subtraction |
| `*` | `a * b` | `50` | multiplication |
| `/` | `a / b` | `0.5` | division (always float) |
| `**` | `a ** 2` | `25` | exponentiation |
| `//` | `16 // 5` | `3` | floor division (drops remainder) |
| `%` | `16 % 5` | `1` | modulo (the remainder itself) |

**Compound assignment** shorthands update a variable in place: `a += 5` is the same as `a = a + 5`.


```python
a, b = 5, 10
print('a + b  =', a + b)
print('a - b  =', a - b)
print('a * b  =', a * b)
print('a / b  =', a / b)
print('a ** 2 =', a ** 2)
print('16 // 5=', 16 // 5)
print('16 %% 5 =' % (), 16 % 5)       # %% to escape the % in an old-style format

# Compound assignment:
x = 10
x += 5        # x = x + 5   -> 15
x *= 2        # x = x * 2   -> 30
print('x =', x)
```

## 4. Comparisons and logical operators

Comparison operators return a **boolean** (`True` or `False`). Note that `==` tests equality (two equals signs!) while a single `=` *assigns*.

| Operator | Reads as |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>`, `<` | greater / less than |
| `>=`, `<=` | greater / less than or equal |
| `and`, `or`, `not` | combine or negate booleans |


```python
temperature = 17
print(temperature > 20)                       # False
print(temperature == 17)                      # True
print(temperature != 20)                      # True
print(temperature > 10 and temperature < 25)  # True
print(not (temperature > 25))                 # True
```

Combine everything above into something you'd actually write.


```python
ticket_price = 12
number_of_tickets = 3
service_fee = 5

subtotal = ticket_price * number_of_tickets
total = subtotal + service_fee

print('Subtotal:', subtotal)
print('Total:   ', total)
```

## 6. Control flow — `if` and `for`

Python uses **indentation** (4 spaces) to mark the body of an `if` block or a `for` loop. Jupyter indents automatically after you press `Enter` on a line ending in `:`.

### `if / elif / else`


```python
score = 72

if score >= 80:
    print('A')
elif score >= 70:
    print('B')
elif score >= 60:
    print('C')
else:
    print('F')
```

### `for` loops over a sequence

`range(n)` yields the integers `0, 1, ..., n-1`. `range(a, b)` yields `a, a+1, ..., b-1`. Loops are how we repeat a calculation many times — essential for simulating a model over several periods or scanning over policy parameters.


```python
# Sum the first 10 integers.
total = 0
for i in range(1, 11):
    total += i
print('sum 1..10 =', total)

# Loop over a list of strings.
for city in ['St. Catharines', 'Hamilton', 'Toronto']:
    print('Hello,', city)
```

## 7. Functions

A function packages a chunk of logic behind a name so you can call it many times with different inputs. Define one with `def`; `return` sends a result back to the caller.


```python
def present_value(future_value, r, periods):
    """Return the present value of `future_value` paid `periods` from now at rate `r`."""
    return future_value / (1 + r) ** periods

print(present_value(100, 0.05, 1))   # $100 one year from now at 5%
print(present_value(100, 0.05, 10))  # ...ten years from now at 5%
```

## 8. Importing packages

Python itself is small. The real power comes from **packages** we `import`. Three we'll use again and again:

- `math` — built-in math functions (`sqrt`, `log`, `exp`, ...).
- `numpy` (usually aliased `np`) — fast arrays of numbers.
- `pandas` (aliased `pd`) — tabular data, like an Excel sheet in code.

`import X as Y` gives the package a shorter nickname `Y` to save typing.


```python
import math
import numpy as np
import pandas as pd

print('sqrt(2)      =', math.sqrt(2))
print('e            =', math.e)
print('np array     =', np.array([1, 2, 3]) * 10)
print('pandas vers. =', pd.__version__)
```

## 9. Pandas DataFrames — Excel in code

A `DataFrame` is a table: rows and named columns. You load one from a CSV with `pd.read_csv()`. You can pass a **URL** instead of a local path and pandas will download the file for you — useful because many data agencies (FRED, World Bank, Statistics Canada) publish CSV endpoints.

The example below pulls the US real consumption series (`PCECC96`) directly from the Federal Reserve of St. Louis.


```python
url = 'https://fred.stlouisfed.org/graph/fredgraph.csv?id=PCECC96'
df = pd.read_csv(url)

# First five rows:
df.head()
```

### Inspect the shape and column names


```python
print('shape (rows, cols):', df.shape)
print('columns:', list(df.columns))
print('data types:')
print(df.dtypes)
```

### Pick one column and compute summary stats

Square-bracket indexing with a string pulls out a single column as a `Series` (essentially a one-column DataFrame).


```python
series = df['PCECC96']
print('min :', series.min())
print('max :', series.max())
print('mean:', series.mean())
print('std :', series.std())

# Full summary in one shot:
df.describe()
```

## 10. Tidying up: parse dates and rename columns

FRED labels the date column `observation_date` and stores it as plain text. Two small cleanups make life easier:

1. `pd.to_datetime()` converts the text to proper timestamps.
2. `.rename(columns=...)` relabels columns to something human-readable.
3. Setting the date as the DataFrame's **index** (with `set_index`) lets every plot auto-label its x-axis with years.

The cell below folds all three steps together.


```python
df['observation_date'] = pd.to_datetime(df['observation_date'])
df = df.rename(columns={'PCECC96': 'consumption'})
df = df.set_index('observation_date')
df.head()
```

## 11. Your first plot

`.plot()` on a Series or DataFrame draws a quick line chart. Behind the scenes pandas is calling `matplotlib` — the Python plotting library. For most of the course this one-liner is all you need.


```python
import matplotlib.pyplot as plt

ax = df['consumption'].plot(figsize=(9, 4), lw=1.5,
                            title='US real personal consumption expenditures')
ax.set_ylabel('Billions of chained 2017 $')
ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```
